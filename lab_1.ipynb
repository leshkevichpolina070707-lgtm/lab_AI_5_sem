```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
scaler = MinMaxScaler()
```

```python
scaler.fit(df[['popularity']])
```

```python
df['popularity'] = scaler.transform(df[['popularity']])
```

```python
scaler_t = StandardScaler()
```

```python
df['tempo'] = scaler_t.fit_transform(df[['tempo']])
```

```python
df_backup = df.copy()
```

```python
df_four = df.copy()
```

```python
df_four.loc[0:10, 'genre'] = None
df_four.loc[5:25, 'popularity'] = None
df_four.loc[100:120, 'tempo'] = None
```

```python
nan_before = df_four.isnull()
```

```python
nan_before.sum()
```

```python
genre_mode = df_four['genre'].mode()[0]
```

```python
popularity_median = df_four['popularity'].median()
```

```python
tempo_mean = df_four['tempo'].mean()
```

```python
df_four['genre'] = df_four['genre'].fillna(genre_mode)
df_four['popularity'] = df_four['popularity'].fillna(popularity_median)
df_four['tempo'] = df_four['tempo'].fillna(tempo_mean)
```

```python
nan_after = df_four.isnull()
nan_after.sum()
```
